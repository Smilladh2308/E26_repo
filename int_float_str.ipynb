## Integers

An integer is a whole number (no decimal points) 

1. I can't use integers as a variable name
2. If i have assigned a variable name to an integer, the variable type = integer

Functions: 

1. type() : takes variable, and tells what type of variable the variable is 
2. print() : prints the variable
3. int() : makes variable into integer (if possible)
4. float(): makes variable into float (if possible)
5. str(): makes variable into string (if possible)
6. dir() : tells what you can do with your variable 

```python
# Make variablenames for variables 
chips = 2 
cookies = 3
```
